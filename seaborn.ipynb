```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips= sns.load_dataset("tips")
tips
tips.dtypes
tips['day'].unique()
```

```python
sns.pairplot(tips, hue="day")
```

```python
sns.distplot(tips['total_bill'])
```

```python
sns.jointplot(x="total_bill",y="tip", data=tips, kind = 'reg')
```

#categorical plots#

```python
sns.countplot(x="sex", hue="smoker", data=tips)
```

```python
import numpy as np
sns.barplot(x="sex",y="tip", data=tips, estimator= np.median)
```

```python
sns.boxplot(x="tip",y="day", data=tips )
```

```python
sns.violinplot(x="total_bill", y="day", data=tips)
```

```python
sns.stripplot(x="total_bill", y="day", data=tips)
```

```python
sns.swarmplot(x="total_bill", y="day", data=tips)
```

```python
sns.violinplot(x="total_bill", y="day", data=tips)
sns.swarmplot(x="total_bill", y="day", data=tips)
```
